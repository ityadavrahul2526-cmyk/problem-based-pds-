Mean


```python
import statistics

print(statistics.mean([66,34,87,101,154,-32,-34]))
print(statistics.mean([90,-87,-56,43,23,54,15,64]))
print(statistics.mean([25,98,69,56,-3.5,50.56,-56]))
```

Median:

```python
import statistics

print(statistics.median([-78,89,67,-90,48,49,76,-90.87]))
print(statistics.median([30,60,65,78,98,87]))
print(statistics.median([34,76,55,43,30,45,67,87,98,65]))
```

Mode:

```python
import statistics

print(statistics.mode([1,7,9,3,56,]))
print(statistics.mode([9,6,5,2,4,3,-9,45]))
print(statistics.mode(['orange', 'brown', 'liteblue', 'darkred']))
```

Standard Deviation:

```python
import statistics

print(statistics.stdev([1, 3, 5, 7, 9, 11]))
print(statistics.stdev([2, 2.5, 1.25, 3.1, 1.75, 2.8]))
print(statistics.stdev([-11, 5.5, -3.4, 7.1]))
print(statistics.stdev([85, 98, 59, 98]))
```

Variance:

```python
import statistics

print(statistics.variance([1, 3, 5, 7, 9, 11]))
print(statistics.variance([2, 2.5, 1.25, 3.1, 1.75, 2.8]))
print(statistics.variance([-11, 5.5, -3.4, 7.1]))
print(statistics.variance([1, 30, 50, 100]))
```
